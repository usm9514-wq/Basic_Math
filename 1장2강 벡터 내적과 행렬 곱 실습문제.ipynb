#### 문제 설명

| 고객 | 방문 횟수 | 장바구니 추가 횟수 |
|---|---:|---:|
| A | 3 | 2 |
| B | 5 | 1 |
| C | 2 | 4 |
| 가중치 | 2 | 4 |

입력과 가중치의 순서는 **방문 횟수 → 장바구니 추가 횟수**입니다. 이번 문제는 편향을 더하지 않고 내적만 계산합니다.

$$x\cdot w=x_1w_1+x_2w_2$$

- $x_1$, $x_2$: 고객의 첫 번째·두 번째 입력값
- $w_1$, $w_2$: 같은 위치의 특성에 적용할 가중치
- $\cdot$: 같은 위치끼리 곱한 값을 모두 더하는 내적 기호

#### 수식 읽는 순서

`x₁w₁`은 ‘첫 입력값 × 첫 가중치’입니다. 두 글자를 붙여 썼지만 곱셈을 뜻합니다. `x₂w₂`도 같은 방식입니다. `x·w`는 이 곱셈 결과들을 모두 더한 값입니다.

‘대응하는 원소’는 **같은 순서에 있는 숫자끼리**라는 뜻입니다. 방문 횟수에는 방문 가중치를, 장바구니 횟수에는 장바구니 가중치를 곱하세요.

### 문제 1-2. 내적의 부호와 코사인 유사도 이해하기

#### 문제 설명

교안의 간단한 방향 예시를 사용합니다. 아래의 모든 벡터는 0벡터가 아닙니다.

| 구분 | 벡터 |
|---|---|
| 기준 벡터 `a` | [1, 0] |
| 같은 방향 `b_same` | [2, 0] |
| 수직 방향 `b_perp` | [0, 2] |
| 반대 방향 `b_opposite` | [−2, 0] |

추가로 교안의 `u = [3, 4]`, `v = [6, 8]`의 코사인 유사도를 계산합니다.

$$\mathrm{cosine\ similarity}=\frac{u\cdot v}{\|u\|_2\times\|v\|_2}$$

분자는 두 벡터의 내적, 분모는 두 벡터의 L2 Norm을 곱한 값입니다.

| 결과 | 방향 관계 |
|---|---|
| 코사인 유사도 1 | 같은 방향 |
| 코사인 유사도 0 | 서로 수직 |
| 코사인 유사도 −1 | 반대 방향 |

> 교안의 코사인 유사도 목록 중 ‘반대 방향: 1’은 ‘반대 방향: −1’의 부호 오타이므로 바로잡아 적용했습니다.

#### 먼저 이해하기

- 벡터를 원점에서 출발하는 화살표로 생각하면, ‘방향’은 화살표가 향하는 쪽입니다.
- `[1,0]`과 `[2,0]`은 같은 쪽, `[1,0]`과 `[0,2]`는 직각, `[1,0]`과 `[−2,0]`은 서로 반대쪽을 향합니다.
- 90°는 직각, 180°는 완전히 반대 방향입니다. 각도를 구하는 코드는 작성하지 않고 표와 내적의 부호를 연결합니다.
- **분자**는 분수의 위쪽, **분모**는 아래쪽입니다. 코사인 유사도 식은 ‘내적 ÷ (첫 길이 × 둘째 길이)’로 읽습니다.
- L2 길이는 이전 강의처럼 ‘각 원소 제곱 → 더하기 → 제곱근’으로 구합니다. 예를 들어 √25는 5×5=25이므로 5입니다.

1~2번은 방향별 내적, 3~4번은 u·v의 코사인 유사도입니다. 두 부분을 차례로 진행하세요.

m은 고객 수, n은 입력 특성 수, p는 출력 수로 읽을 수 있습니다. 가운데 크기가 같아야 계산할 수 있습니다.

결과의 행 수는 왼쪽 행렬에서, 결과의 열 수는 오른쪽 행렬에서 가져옵니다. 이 문제에서는 **고객 수 × 출력 점수 종류 수**입니다.

1~2번은 고객별 한 점수, 3~4번은 고객별 두 점수, 5번은 Shape를 읽는 연습입니다. 5번은 실제 배열을 만들지 않고 숫자의 개수만 비교하세요.

#### 답변 작성란

**수식 및 계산 과정**

- 작성: 

**Q1. X와 W의 Shape는 무엇이며 행렬곱이 가능한 이유는 무엇인가요?**

- 답변: X는 (2, 2), W는 (2, 2)입니다. X의 열 수(2)와 W의 행 수(2)가 같아서 행렬곱이 가능합니다.
