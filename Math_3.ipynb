로그
어떤 밑을 제곱하면 이 값이 되는가를 알려주는 연산자

예약 세건의 실제 결과에 확률이 0.8, 0.7 ,0.7 이라고 가정을 해보면

우도는 이 세값을 곱한 0.392 값을 의미한다.

이 우도의 자연 로그는 ln(0.392)로 구할 수 있고 ln(0.8) + ln (0.7) + ln (0.7)로 얻을 수 있다.

즉 , 로그의 특성을 이요하면 확률을 모두 곱하는 과정을 확률의 로그를 더하는 과정으로 바꿀 수 있다.

기록이 매우 많아서 계속 곱해서 0에 한도 끝도 없이 가까워지면 구분하기 어려우니까

같은 비교지만 로그를 써서 안정적으로 계산하겠다는 의미이다.



log likelihood

likelihood 에 자연 로그를 적용한 값

각 관측 확률의 자연로그를 더해서도 같은 값을 얻는다

같은 데이터에서는 큰 log likelihood를 선호한다.

q = [0.8, 0.7 ,0.7] 이면 L = 0.392고, ln(L) = ln(0.8) + ln(0.7) + ln(0.7) = -0.936

코드에서는 np.sum(np.log(q))를 사용할 수 있다.

예를 들어 다른 모델이 같은 세 결과에 각각 0.5를 주면 liklihood는 0.125

로그 우도는 약 -2.0794 값

-0.9365 > -2.0794 이기 때문에 첫 모델을 선택하려고 한다.

우도로 비교해도 0.392 > 0.125 보다는 크기 때문에 같은 선택을 한다.

각 주문에서 실제 시간과 예츠,ㄱ 시간의 차이를 제곱하면 [4, 4, 9] 다 <- 평균을 계산해보면 17/3 = 5.66

같은 주문에서 B의 평균 제곱 오차가 8.667 이라면  5.66인 a모델이 더 좋았다고 평가 할 수 있다.

손실 함수는 모델의 예측을 비교하고 학습중 어떤 것을 줄여야 하는지  정하는 역활을 한다.





 SSE 와 제곱 오차합

 제곱 오차를 모두 더한 값

 제곱 오차ㅡ[4, 4,9]의 SSE 는 17 이다.


 배달 3건에 대해서 A 모델의 제곱 오차가 [ 4,4,9 ] , B의 제곱 오차가 [1 , 16, 9] 라고 가정해보면

 A의 합계는 17, B는 26 <- 이 합계가 의미하는 것이 SSE

 같은 세건에 대해서는 A의 전체 제곱오차가 더 작다.

 B가 비록 첫 주문을  A 모델보다 더 잘 맞췄더라도 전체 합계에서는 A가 더 좋은 모델일 수도 있다.

 SSE를 전체 주문 건수로 나누면 MSE가 된다.

 그러면 이걸 왜 하느냐?


SSE를 통해 MSE를 구할 수 잇고, 합계가 커진 이유 자체가 오차인지, 데이터 의 수에 영향을 받았는지 구분할 수 있다.



평균 제곱 오차 MSE(mean squared Error)

제곱오차의 평균

각 고객의 오차를 제곱하고 모두 더한 뒤 전체 고객 수로 나눈다.


배달 세건의 실제 시간이 [10, 20, 30] 분이고 예측이 [12, 18 ,27] 분이면

실제 예측은 [-2,2,3 ] , 각 오차의 제곱은 [4 ,4, 9] 합계 17을 3(전체 고객 수 )으로 나누면 5.66 

다른 B 모델의 MSE가 8.66 이라면 A 모델을 서냍ㄱ하는것이 맞다.

그렇다면 왜 중요하게 배우는건가?

여러 수치의 예측 제곱 오차를 하나의 평균으로 요약해 비교하기 위해서다

오차를 전체 수로 평균을 내기 때문에 각 한건에 대한 오차 범위를 대략적으로 파악할 수 있다.




MSE 단위


오차를 제곱하면 단위도 제곱하면 된다.

단위가 만원 단위였다면 제곱 오차의 대한 해석도 (만원)^2 단위로 읽으면 된다.


이진 교차엔트로피와 BCE 

Binary Cross Entropy, 구매/ 미구매 두 가지 결과를 구분하는 분류 손실 함수


한 고객의 BCE는 실제 정답 확률에 음의 로그를 적용한 값

예를 들어보면 

두 예약자에 실제 방문 결과가 [1, 0 ] 이고, A의 방문 확률이 [0.8, 0.3 ] 이라고 가정하면

첫번째 손님의 손실은 -ln(0.8) = 약 0.223 , 둘째는 미방문 이니까 -ln(0.7) = 약 0.356 정도

평균 BCE값은 약 0.289 정도 

B가 두사람 모두에게 방문 확률 0.5 를 주었다면 두 손실은 각각 0.6931로 평균도 0.6931로 나타난다.

같은 두 예약 확률에서는 BCE가 작은 A 모델을 선택한다.

왜 쓰냐?

두 예측 결과를 학습하고 평가할 기준으로 쓴다. 각 정답에 어느정도의 확률을 부여 했는지 알 수 있다.






범주형 Categorical Entrophy

Categorical Entrophy , 여러 종류 중 실제 정답이 각 정답이 하나인 분류에 쓰는 손실

상품 B의  정답이 수박 [배, 수박, 사과] 확률이 [0.2, 0.6, 0.2] 라면 CE = -ln(0.6) = 약 0.511


예를 들어 과일 종류를 [ 배, 수박, 사과] 순서로 분류 한다고 했을 때

첫 문의의 실제 정답은 수박이고 예측 확률이 [0.2, 0.6 , 0.2] 일때 정답인 수박의  0.6을 선택하여 

-ln(0.6) = 약 0.511 을 계산한다.

두번째 정답 확률이 0.8이라면 CE = 0.2231, 두 과일 선택의 평균을 내보면  0.367 정도

왜쓰느냐? 

과일 선택처럼 대상마다 열의 확률을 고르고 대상별 평균 내는 계산을 활용 할 수 있다.



예측 클래스와 손실의 차이

예측 클래스는 모델이 최종적으로 선택한 종류

손실은 실제 정답에 준 확률을 평가한 값


정답을 맞추면 손실이 0이냐? -> 아님

예를 들어

실제 구매 고객에게 확률 0.9와 0.6을 주면 임계값 0.5 에서 둘다 구매로 예측


하지만 BCE는 약 0.105 와 0.511 로 다를 수 있음

즉 -> 정답을 맞췄다는 사실만으로 예측의 품질을 알 수가 없기 때문에 손실 점수를 확인해야 한다.

두가지 사실, 정답을 맞췄는가? / 확률 예측 자체가 얼마나 좋은가? 에 대한 답을 할 수 있다.




정규분포(가우시안)

평균을 중심으로 좌우가 대칭인 종 모양의 연속형 분포

예를 들어 보면 배달 모델에서 주문 도착 시간을 20분으로 예상 했다고 가정하면

실제 배달 시간이 19분, 21분처럼 약간의 오차가 발생할 수는 있다.

비슷한 조건에서 주문에서 작은 차이는 자주, 큰차이는 드물게 나타난다고 설명하기 위해

정규분포를 가정할 수 있다.
