```python
import numpy as np

x = np.array([2, 3])

W = np.array([
    [2, 0],
    [1, 3]
])

b = np.array([1, -2])

```

```python
T = x@W

print(T)

affineline_result = T + b
print(affineline_result)
```

```python

x = np.array([0, 0])

W = np.array([
    [2, 0],
    [1, 3]
])

b = np.array([1, -2])
```

```python
T = x@W

print(T)

affineline_result = T + b
print(affineline_result)
```

선형 변환 (Linear Transformation): 입력값에 행렬을 곱하는 연산(x @ W)입니다. 공간을 회전시키거나, 늘리거나, 찌그러뜨리는 역할을 합니다. 단, 원점(0, 0)을 반드시 지나야 한다는 대전제가 있습니다. 즉, 입력이 모두 0이면 결과도 항상 0이 됩니다.

아핀 변환 (Affine Transformation): 선형 변환에 편향을 더한 연산(x @ W + b)입니다. 선형 변환으로 공간을 변형시킨 뒤, 편향(bias)을 이용해 원점 위치에서 통째로 평행이동(Shift)시키는 확장된 개념입니다.

```python
X = np.array([[3, 2], [5, 1]])

w = np.array([[2], [-1]])
b = 3

T = X@w + b
print(T)

```
