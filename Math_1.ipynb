```python
Scaler
it refers to the allocating a numbers into specific words

vector
the linear representation of data in a multi-dimensional space, often used in machine learning and data analysis to represent features or attributes of data points.

vector dimension
the number of components or features in a vector, indicating the size of the vector space in which the data points are represented.
 For example, a vector with three components (x, y, z) has a dimension of 3.

shape 
it shows how many rows and columns are in a data structure, such as an array or matrix. 
 x.shape = (2, ) 

row and column vectors
Row vectors are represented as a single row of data, while column vectors are represented as a single column of data. 
In mathematical notation, a row vector is typically written as a 1 x n matrix, while a column vector is written as an n x 1 matrix.

reshape
The reshape operation allows you to change the shape of an array or tensor without altering its data. 
This is useful for converting between different representations of the same data, such as turning a 1D array into a 2D matrix or vice versa.
x=np.array([1, 2, 3, 4, 5, 6])
x.reshape(2, 3)


weights
weights are numerical values assigned to features in a model, indicating their importance or contribution to the prediction or output. 
In machine learning, weights are often learned during the training process to optimize the model's performance.


bias
bias is an additional parameter in a model that allows it to make predictions even when all input features are zero.


parameter
파라미터는 모델의 계산에 사용되며 학습으로 조정할 수 있는 값
예를 들어 공부 시간으로 시험 점수를 예상한다고 가정해보면

예상점수 = 공부 시간 x 5 + 10 <- 이라는 기본 규칙에서

실제 2시간 공부한 학생의 예상점수는 20점인데, 실제로  30점이라면 두 점수 차이를 계산하여
이를 조정하는 값

parameter refers to the adjustable values in a model that are learned from the training data.
 These parameters are optimized during the training process to minimize the error between the predicted and actual outcomes.
  For example, in a linear regression model, the slope and intercept are parameters that can be adjusted based on the training data to improve predictions.

가중합과 출력 점수
가중합은 각 입력값에 대응하는 가중치를 곱한 뒤 모두 더한 값

아까 앞서 5 x 2 = 10점 , 2 x 3 = 6점, 편향 1점 이면 17점

-> 여기서 16점이 가중합이라고 부를 수 있다.

가중치 행렬

가중치 행렬은 여러 출력에 필요한 가중치들을 표로 모은것

W = [[2,1], [3,4]] 라고 가정하면 같은 출석, 과제 기록으로 점수를 만든다고 할때

첫번째 점수는 출석당 2점, 과제당 3점, 둘째 점수는 출석당 1점, 과제당 4점으로 계산된다.



노름 L1 Norm
L1 노름은 각 원소의 절대값을 구한 뒤 전부 더한 값
윤하가 오른쪽으로 3m 이동한뒤 아래쪽으로는 4m 이동했다면, L1 Norm은 3, -4 일수 있다.

이걸 이동 거리로 계산해보면 |3| + |-4| = 7m 이동한것으로 계산된다.


L2 Norm
L2 norm은 각 원소의 제곱을 구한 뒤 전부 더하고, 마지막으로 제곱근을 구한 값
윤하가 오른쪽으로 3m 이동한뒤 아래쪽으로는 4m 이동했다면, L2 Norm은 √(3² + (-4)²) = √(9 + 16) = √25 = 5m 이동한것으로 계산된다.


차이 백터
차이 벡터는 두 벡터 간의 차이를 나타내는 벡터로, 각 요소는 해당 위치의 두 벡터 요소 간의 차이이다.
 For example, if we have two vectors A = [a1, a2, a3] and B = [b1, b2, b3], the difference vector D = A - B would be calculated as D = [a1 - b1, a2 - b2, a3 - b3].


manhattan distance
manhattan distance, also known as L1 distance or taxicab distance, is a measure of distance between two points in a grid-based system.
 It is calculated as the sum of the absolute differences of their coordinates. 

 For example, if point A has coordinates (x1, y1) and point B has coordinates (x2, y2), the Manhattan distance is |x1 - x2| + |y1 - y2|.


euclidean distance
euclidean distance is a measure of the straight-line distance between two points in Euclidean space. 
 It is calculated using the Pythagorean theorem, which involves taking the square root of the sum of the squared differences of their coordinates. 

 For example, if point A has coordinates (x1, y1) and point B has coordinates (x2, y2), the Euclidean distance is √((x1 - x2)² + (y1 - y2)²).



vector direction and ratio
direction of a vector refers to the orientation of the vector in space, which can be represented by the angle it makes with a reference axis or by its components in a coordinate system. 
The ratio of a vector can refer to the relative proportions of its components, which can be used to describe its direction and magnitude.
a = np.array([3, 4]) b = np.array([6, 8]) b is twice the length of a, but both vectors have the same direction since they are scalar multiples of each other. 
The ratio of their components is 1:2, indicating that b is in the same direction as a but scaled by a factor of 2.
```
