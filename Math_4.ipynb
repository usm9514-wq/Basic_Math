미분은 지금 위치의 변화율을 나타냄

|구간| 출력 변화량|평균 변화율|

|구간| 출력 변화량|평균 변화율|
|---|---|---|
|3 -> 4| 1| (16-9)/ 1 =7|
|3 -> 3.1| 0.1 | (9.61- 9) / 0.1 = 6.1|
|3 -> 3.01| 0.01 | (9.0601 -9) / 0.01 = 6.01|
|3 -> 3.001| 0.001 | (9.00601 -9) / 0.001 = 6.001|

구간을 줄일 수록 변화율이 6에 가까워짐

x = 3 에서 순간 변화율은 6이고, 이렇게 한 점을 찍었을 때 순간 변화율을 구하는것이 미분임

+5는 그래프의 전체 높이를 높일 뿐 기울기를 바꾸지 않는다.

각 중간값에 대한 손실의 기울기를 계산하여 역전파에서 전달

손실에서 시작한다면 어떤것을 미분 해야하는가?
예측값이 조금 달라지면 손실은 어떻게 달라지는가?
가중치 편향 고려하지 않고 y_pred 의 값이 조금 바뀐다고 가정해보면

|예측값| 손실 계산| 손실|

예측을 0.01 늘렸더니 손실은 0.059 감소, 현재 기울기 - 6 으로 예상한 변화 -6 x 0.01에 가까운 수치

가중치와 평향까지 기울기 전달하기

앞서 기울기 -6, ∂L/∂ŷ=−6 의 계산식을 얻었다

ŷ = wx + b 에서 w와 b가 각각 무엇을 바꿀 수 있는 지? ( x = 2 고정)
가중치 w를 0.01 늘리면 예측이 2 x 0.01 = 0.02 늘어남

따라서 w에 대한 예측의 변화율은 2, 편향 b도 0.01 늘리면 예측도 0.01 늘어나기 때문에 변화율은 1 

∂ŷ/∂w = x = 2 / ∂ŷ/∂b = 1

-6 x 2  = - 12 와 -6 x 1 - -6 값을 얻을 수 있는데

|현재 계산 대상| 값| what|

| ∂L/∂b|-6 | b를 조금 바꿀때 손실의 변화율|
