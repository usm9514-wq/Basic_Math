### 문제 1-2. 입력·정답·가중치·편향 구분과 점수 계산

#### 문제 설명

문제 1-1과 같은 고객 데이터에 실제 구매 여부를 추가했습니다.

| 고객 | 방문 횟수 | 장바구니 추가 횟수 | 실제 구매 여부 |
|---|---:|---:|---:|
| A | 5 | 2 | 1 |
| B | 3 | 1 | 0 |
| C | 8 | 4 | 1 |

구매한 경우는 1, 구매하지 않은 경우는 0입니다. 방문 횟수의 가중치는 2, 장바구니 추가 횟수의 가중치는 3, 편향은 1로 정합니다.

교안의 계산식을 그대로 사용합니다.

$$\mathrm{score}=w_1x_1+w_2x_2+b$$

| 기호 | 의미 |
|---|---|
| $x_1$, $x_2$ | 방문 횟수, 장바구니 추가 횟수 |
| $w_1$, $w_2$ | 각 특성에 곱하는 가중치 |
| $b$ | 마지막에 더하는 편향 |
| score | 가중합으로 계산한 출력 점수 |

또한 교안의 두 출력용 가중치 표를 행렬로 표현해 보세요.

| 입력 특성 | 구매 관심도 가중치 | 고객 활동 가중치 |
|---|---:|---:|
| 방문 횟수 | 2 | 1 |
| 장바구니 추가 횟수 | 3 | 4 |

#### 수식을 말로 읽기

`w₁x₁`은 **첫 가중치 × 첫 입력값**입니다. 두 글자가 붙어 있어도 곱셈을 뜻합니다. 아래첨자 1과 2는 첫 번째·두 번째를 구분하는 번호입니다.

전체 식은 ‘방문 횟수에 해당 가중치를 곱한다 → 장바구니 추가 횟수에도 해당 가중치를 곱한다 → 두 결과와 편향을 더한다’는 뜻입니다. **가중합**은 이렇게 ‘각각 곱한 뒤 합한 값’입니다. **편향**은 그 합에 마지막으로 더하는 값입니다.

4. 두 출력용 표를 가중치 행렬 `W`로 만들고 배열과 Shape를 출력하세요. 행은 입력 특성, 열은 출력 점수에 대응하도록 배치하세요. 이 문항에서는 W를 이용한 출력 계산은 하지 않습니다.

#### 작성 도움말

```python
# 문제 1-2. 입력·정답·가중치·편향 구분과 점수 계산
# 요구사항을 한 단계씩 작성하고 실행하세요.
# 계산 방법과 함수 이름은 위의 작성 도움말을 참고하세요.

import numpy as np

X = np.array([[5, 2], [3, 1], [8, 4]])
y = np.array([1, 0, 1])
w = np.array([2, 3])
b = 1
W = np.array([[2, 1],
              [3, 4]])

print('입력 행렬', X)
print('정답 벡터', y)
print('가중치 벡터', w)
print('편향 스칼라', b)
print('x.shape', X.shape)
print('y.shape', y.shape)
print('w.shape', w.shape)


# Q3
score_a = X[0] @ w + b
print(score_a)          # 17

# Q4
score = X @ w + b
print(score)            # [17 10 29]  ← y는 안 쓰임, 업데이트 없음

# Q5
print(W[:, 0])          # [2 3]  구매 관심도 가중치
print(W[:, 1])          # [1 4]  고객 활동 가중치
print(X @ W)            # [[16 13] [ 9  7] [28 24]]
```

#### 문제 설명

교안의 벡터 `v = [3, -4]`를 사용합니다. 이 벡터는 노름 계산을 연습하기 위한 숫자 예시입니다.

노름은 벡터의 크기를 하나의 숫자로 표현합니다. 두 원소에 대한 공식은 다음과 같습니다.

$$\|v\|_1=|v_1|+|v_2|$$
$$\|v\|_2=\sqrt{v_1^2+v_2^2}$$

- $v_1$, $v_2$: 벡터의 첫 번째와 두 번째 원소
- $|a|$: a의 절댓값. 예를 들어 −4의 절댓값은 4입니다.
- $a^2$: a를 자기 자신과 곱한 값
- $\sqrt{a}$: 제곱하여 a가 되는 수 중 음수가 아닌 값. 예를 들어 $\sqrt{25}=5$입니다.
- $\|v\|_1$, $\|v\|_2$: 각각 벡터 v의 L1 Norm, L2 Norm

#### 수식을 말로 읽기

- `||v||₁`은 ‘벡터 v의 엘원 노름’, `||v||₂`는 ‘벡터 v의 엘투 노름’으로 읽습니다. 아래의 1과 2는 **계산 방법의 종류**입니다.
- 절댓값은 숫자가 0에서 얼마나 떨어져 있는지 나타냅니다. 양수는 그대로이고 음수는 크기만 남깁니다.
- 제곱은 같은 수를 두 번 곱하는 것입니다. 음수 전체를 제곱할 때는 `(-4)²`처럼 괄호로 묶어 읽으세요.
- 제곱근 기호 √는 ‘어떤 음수가 아닌 수를 두 번 곱해야 이 값이 되는가’를 묻습니다.

### 문제 2-2. 두 고객 사이의 거리와 벡터 정규화

#### 문제 설명

교안의 연습문제와 같은 두 고객의 활동 데이터입니다.

| 고객 | 방문 횟수 | 장바구니 추가 횟수 |
|---|---:|---:|
| A | 3 | 4 |
| B | 6 | 8 |

원소 순서는 방문 횟수 → 장바구니 추가 횟수입니다.

두 고객 사이의 거리는 **B 벡터 − A 벡터**에 노름을 적용합니다.

$$d_1(A,B)=\|x_B-x_A\|_1,\qquad d_2(A,B)=\|x_B-x_A\|_2$$

길이가 1인 벡터를 만들 때는 각 벡터를 **자신의 L2 Norm**으로 나눕니다.

$$x_{\mathrm{normalized}}=\frac{x}{\|x\|_2}$$

이 방법은 L2 Norm이 0이 아닌 벡터에 적용합니다.

#### 수식을 말로 읽기

`x_B−x_A`는 **같은 위치의 특성끼리 뺀다**는 뜻입니다. 방문 횟수끼리, 장바구니 추가 횟수끼리 뺍니다. `d₁(A,B)`와 `d₂(A,B)`는 A와 B 사이의 거리를 두 방법으로 계산한 값입니다.

정규화 식의 분수는 ‘위에 있는 벡터의 **모든 원소**를 아래의 L2 Norm 값으로 나눈다’고 읽습니다. 길이가 1이라는 뜻은 **L2 Norm이 1**이라는 뜻입니다.

**1~3번은 거리**, **4~5번은 원래 길이를 구한 뒤 정규화**, **6~7번은 결과 확인**입니다. 한 묶음씩 진행하세요.
